# GeoPandas-AI: Real-Time Interactive Chat

Run cells sequentially to have a conversation with your data.
Each cell builds on the previous one!

```python
# Setup (run once)
import geopandas as gpd
from geopandasai import GeoDataFrameAI, update_geopandasai_config
import folium
from matplotlib.figure import Figure
import warnings
warnings.filterwarnings('ignore')

# Configure Ollama
update_geopandasai_config(
    lite_llm_config={
        "model": "openai/qwen2.5-coder:14b",
        "api_base": "http://127.0.0.1:11434/v1",
        "api_key": "ollama",
    }
)

# Load your data
gdf = gpd.read_file("admin_boundaries.geojson")
gdf_sample = gdf.sample(n=50, random_state=42)

# Create AI-enabled GeoDataFrame
gdfai = GeoDataFrameAI(gdf_sample)

print(f"✓ Ready! Loaded {len(gdf_sample)} features")
print("Run cells below to chat with your data")
```

---
## Start Your Conversation

Edit the prompt below and run the cell:

```python
# 🗨️ YOUR PROMPT HERE - Edit and run!
YOUR_PROMPT = "Create an interactive map showing the districts"

result = gdfai.chat(YOUR_PROMPT, return_type=folium.Map)

# Display inline in Jupyter
result
```

---
## Improve It!

Run this cell multiple times with different improvements:

```python
# 🗨️ IMPROVEMENT PROMPT - Change and re-run!
IMPROVEMENT = "Add colors for each district"

result = gdfai.improve(IMPROVEMENT, return_type=folium.Map)

# Display the improved map
result
```

```python
# 🗨️ Keep improving!
IMPROVEMENT = "Add tooltips showing district names"

result = gdfai.improve(IMPROVEMENT, return_type=folium.Map)
result
```

```python
# 🗨️ One more improvement
IMPROVEMENT = "Make the colors more vibrant"

result = gdfai.improve(IMPROVEMENT, return_type=folium.Map)
result
```
